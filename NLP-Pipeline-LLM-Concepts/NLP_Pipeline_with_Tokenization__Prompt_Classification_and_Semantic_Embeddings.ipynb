###  1. Import & Setup

```python
from transformers import AutoTokenizer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

### 2. Create the Prompt Dataset



```python
# Unified dataset

sentences = [
    # Education
    "Artificial Intelligence is transforming education.",
    "Students learn better with technology.",
    "Universities should use AI to improve learning.",
    "Teachers benefit from AI-powered tools.",

    # Technology
    "Deep learning helps prediction.",
    "AI will change transportation.",
    "Self-driving cars rely on AI.",
    "Machine learning improves automation.",

    # Healthcare
    "Doctors treat patients in hospitals.",
    "Medical professionals diagnose illnesses.",
    "AI helps doctors make better decisions.",
    "Hospitals use AI for diagnosis.",

    # Food
    "Pizza tastes good.",
    "I enjoy eating Italian food.",
    "Spaghetti with cheese is delicious.",
    "Burgers and fries are popular fast food.",
]

labels = [
    "education",
    "education",
    "education",
    "education",

    "technology",
    "technology",
    "technology",
    "technology",

    "healthcare",
    "healthcare",
    "healthcare",
    "healthcare",

    "food",
    "food",
    "food",
    "food"
]

print("Unified dataset loaded. Total samples:", len(sentences))
```

###  3. Tokenization + Classical TF‑IDF Classification


```python
# Tokenization demo
tokenizer = AutoTokenizer.from_pretrained("gpt2")
demo_sentence = "Artificial Intelligence is transforming education."
print("Tokens:", tokenizer.tokenize(demo_sentence))

# TF-IDF with n-grams
vectorizer_B = TfidfVectorizer(ngram_range=(1, 2))
X_B = vectorizer_B.fit_transform(sentences)

clf_B = LogisticRegression()
clf_B.fit(X_B, labels)

# User input
user_input_B = input("\nEnter a sentence for Part B classification: ")
vec_B = vectorizer_B.transform([user_input_B])

probs_B = clf_B.predict_proba(vec_B)[0]
pred_B = clf_B.predict(vec_B)[0]
conf_B = max(probs_B)

print(f"Predicted category (Part B): {pred_B} (confidence: {conf_B:.3f})")
```

### 4. Prompt Classification

```python
print("\n=== PART C: PROMPT CLASSIFICATION ===")

# Reuse unified dataset
vectorizer_C = TfidfVectorizer(ngram_range=(1, 2))
X_C = vectorizer_C.fit_transform(sentences)

clf_C = LogisticRegression()
clf_C.fit(X_C, labels)

user_input_C = input("\nEnter a prompt for Part C classification: ")
vec_C = vectorizer_C.transform([user_input_C])

probs_C = clf_C.predict_proba(vec_C)[0]
pred_C = clf_C.predict(vec_C)[0]
conf_C = max(probs_C)

print(f"Predicted category (Part C): {pred_C} (confidence: {conf_C:.3f})")
```

###  5. Embedding‑Based Semantic Classification

```python
model_D = SentenceTransformer('all-MiniLM-L6-v2')

emb_D = model_D.encode(sentences)

clf_D = LogisticRegression()
clf_D.fit(emb_D, labels)

user_input_D = input("\nEnter a sentence for Part D semantic classification: ")
user_emb = model_D.encode([user_input_D])

pred_D = clf_D.predict(user_emb)[0]
probs_D = clf_D.predict_proba(user_emb)[0]
conf_D = max(probs_D)

print(f"Predicted category (Part D): {pred_D} (confidence: {conf_D:.3f})")

# Semantic similarity
sims = cosine_similarity(user_emb, emb_D)[0]
best_idx = np.argmax(sims)
best_sim_label = labels[best_idx]
best_sim_score = sims[best_idx]

print(f"Most similar training sentence label: {best_sim_label} (similarity: {best_sim_score:.3f})")
```
